# Sample problem (an example, not for submission)

The table `trails` lists some Humboldt hiking trails, their length in miles, and whether dogs are allowed (made-up values for this problem):

| Name | Miles | Dogs |
|---|---|---|
| Hammond Trail | 5.5 | True |
| Arcata Marsh Loop | 2 | True |
| Trinidad Head Loop | 1.4 | True |
| Fern Canyon Loop | 1.1 | False |
| Rhododendron Trail | 6.3 | False |
| Elk Camp Ridge | 3.8 | True |

Which expression evaluates to the **name of the longest trail that allows dogs**? Circle one.

(a) `trails.where('Dogs', True).sort('Miles', descending=True).column('Name').item(0)`

(b) `trails.sort('Miles', descending=True).column('Name').item(0)`

(c) `trails.where('Dogs', True).sort('Miles').column('Name').item(0)`

(d) `trails.where('Dogs', True).group('Name', max).column('Name').item(0)`

```python
# Setup code: builds the table so you can check the answer
trails = Table().with_columns(
    'Name', make_array('Hammond Trail', 'Arcata Marsh Loop', 'Trinidad Head Loop',
                       'Fern Canyon Loop', 'Rhododendron Trail', 'Elk Camp Ridge'),
    'Miles', make_array(5.5, 2.0, 1.4, 1.1, 6.3, 3.8),
    'Dogs', make_array(True, True, True, False, False, True))
trails
```

## Sample solution

**(a)**, which gives `'Hammond Trail'`. It keeps only the dog-friendly rows, sorts them longest first, and takes the first name.

Why the others are wrong:
- **(b)** never filters on `Dogs`, so it returns the longest trail overall, `'Rhododendron Trail'` (no dogs allowed).
- **(c)** sorts in ascending order, so it returns the **shortest** dog-friendly trail, `'Trinidad Head Loop'`.
- **(d)** `group('Name', max)` makes one row per name, sorted **alphabetically**, so `item(0)` is `'Arcata Marsh Loop'`. Grouping doesn't sort by miles.

A good question has one clearly correct answer, and its wrong options each come from a mistake a classmate might really make.

```python
# Check: run each option
print('(a)', trails.where('Dogs', True).sort('Miles', descending=True).column('Name').item(0))
print('(b)', trails.sort('Miles', descending=True).column('Name').item(0))
print('(c)', trails.where('Dogs', True).sort('Miles').column('Name').item(0))
print('(d)', trails.where('Dogs', True).group('Name', max).column('Name').item(0))
```
