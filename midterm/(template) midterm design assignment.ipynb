# Sample problem (an example, not for submission)

The table `trails` lists some Humboldt hiking trails (made-up values):

| Name | Miles | Dogs |
|---|---|---|
| Hammond Trail | 5.5 | True |
| Arcata Marsh Loop | 2 | True |
| Trinidad Head Loop | 1.4 | True |
| Fern Canyon Loop | 1.1 | False |
| Rhododendron Trail | 6.3 | False |
| Elk Camp Ridge | 3.8 | True |

Write one line of code that evaluates to the **name of the longest trail that allows dogs**. What value does it give?

```python
# Setup code (optional): builds the table so you can check your answer
trails = Table().with_columns(
    'Name', make_array('Hammond Trail', 'Arcata Marsh Loop', 'Trinidad Head Loop',
                       'Fern Canyon Loop', 'Rhododendron Trail', 'Elk Camp Ridge'),
    'Miles', make_array(5.5, 2.0, 1.4, 1.1, 6.3, 3.8),
    'Dogs', make_array(True, True, True, False, False, True))
trails
```

## Sample solution

`trails.where('Dogs', True).sort('Miles', descending=True).column('Name').item(0)` gives **'Hammond Trail'**.

Keep only the trails that allow dogs, sort them longest first, then take the first name. (Without the `where`, you'd get Rhododendron Trail, which doesn't allow dogs.)

```python
trails.where('Dogs', True).sort('Miles', descending=True).column('Name').item(0)
```
