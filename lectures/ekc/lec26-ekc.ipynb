```python

    'Mean of 900 flight delays', sample_means_900).hist(bins=20)

```

</details>

---
> **Today's takeaway (the CLT):** means of large random samples are approximately normal, centered at the population mean, with $$SD = \frac{\text{population SD}}{\sqrt{n}}$$ *no matter what the population looks like*!!!

### The Central Limit Theorem
```python
# distribution of the sample mean:
#   center  = population mean
#   SD      = population SD / sqrt(sample_size)
#   shape   = approximately normal (for large samples)
```
